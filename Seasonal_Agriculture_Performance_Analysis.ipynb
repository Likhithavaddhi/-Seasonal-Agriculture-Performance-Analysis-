```python
from scipy import stats

print("Statistical library loaded successfully!")
```

```python
kharif = data[data['Season'] == 'Kharif']['Yield_Tonnes_Ha'].dropna()
rabi = data[data['Season'] == 'Rabi']['Yield_Tonnes_Ha'].dropna()
zaid = data[data['Season'] == 'Zaid']['Yield_Tonnes_Ha'].dropna()

print("Kharif records:", len(kharif))
print("Rabi records:", len(rabi))
print("Zaid records:", len(zaid))
```

```python
anova_result = stats.f_oneway(kharif, rabi, zaid)

print("F-statistic:", round(anova_result.statistic, 4))
print("P-value:", round(anova_result.pvalue, 6))
```

```python
alpha = 0.05

if anova_result.pvalue < alpha:
    print("Result: There is a statistically significant difference in yield among the seasons.")
else:
    print("Result: There is no statistically significant difference in yield among the seasons.")
```

```python
statistical_result = pd.DataFrame({
    'Test': ['One-way ANOVA'],
    'Variable': ['Yield_Tonnes_Ha'],
    'Groups': ['Kharif, Rabi, Zaid'],
    'F_Statistic': [anova_result.statistic],
    'P_Value': [anova_result.pvalue],
    'Significance_Level': [0.05]
})

statistical_result.round(6)
```

```python
from scipy import stats

season_groups = [
    group['Yield_Tonnes_Ha'].dropna()
    for _, group in data.groupby('Season')
]

f_stat, p_value = stats.f_oneway(*season_groups)

print("F-statistic:", round(f_stat, 4))
print("p-value:", round(p_value, 6))
```

```python
alpha = 0.05

if p_value < alpha:
    print("Result: There is a statistically significant difference in average yield across seasons.")
else:
    print("Result: There is no statistically significant difference in average yield across seasons.")

print("Significance level (α):", alpha)
print("p-value:", round(p_value, 6))
```
